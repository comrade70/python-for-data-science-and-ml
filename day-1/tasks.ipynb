# Function 
A python function called “two_sum” that takes in two arguments(integer) and calculates the sum of two number


```python
def two_sum(a, b):
    return a + b
integer_numbers = two_sum(2, 3)
print(integer_numbers)
```

# If-else statement 
An if-else statement that print “This is a prime number” if the number is prime, or print out “This is an Ordinary number” if the number is not prime


```python
number = int(input("Enter an integer number "))
if number > 1:
    for i in range(2, number):
        if number % i == 0:
            print("This is an Ordinary number")
            break
            else:
        print("This is a prime number")
else:
    print("This is an Ordinary number")
```
